# Exercise 1.B NumPy Random

```python
# Import NumPy
import numpy as np
```

## Generating Random Numbers

```python
# Generate 2 random integers between 1 and 20
r_1 = np.random.randint(1, 21, size=2)

print(r_1)
```

```python
# Generate 1 row and 2 columns of random integers
r_2 = np.random.randint(1, 21, size=(1, 2))

print(r_2)
```

```python
# Generate 2 rows and 1 column of random floats between 0 and 1
r_3 = np.random.random(size=(2, 1))

print(r_3)
```

```python
# Generate 4 rows and 5 columns of random integers between 20 and 100
r_4 = np.random.randint(20, 101, size=(4, 5))

print(r_4)
```

```python
# Generate 4 rows and 5 columns of random floats between 0 and 1
r_5 = np.random.random(size=(4, 5))

print(r_5)
```

## Generating Random Choices from an Array

```python
# Create an array containing 12 classmates
students = np.array([
    "Student1",
    "Student2",
    "Student3",
    "Student4",
    "Student5",
    "Student6",
    "Student7",
    "Student8",
    "Student9",
    "Student10",
    "Student11",
    "Student12"
])

print(students)
```

```python
# Randomly select one student
call_on = np.random.choice(students)

print(call_on)
```

```python
# Randomly arrange students into 3 groups with 4 students each
breakout_groups = np.random.choice(students, size=(3, 4), replace=False)

print(breakout_groups)

# This is an effective way to assign breakout groups because
# students are randomly selected and nobody is assigned twice.
```

```python
# Randomly select five students using different probabilities
call_on_weighted = np.random.choice(
    students,
    size=5,
    p=[0.30, 0.15, 0.10, 0.10, 0.10, 0.10,
       0.05, 0.05, 0.05, 0.00, 0.00, 0.00]
)

print(call_on_weighted)

# Student 1 has a 30% chance of being selected.
# Students 2-9 have chances between 5% and 15%.
# Students 10-12 have a 0% chance of being selected.
# The probabilities add up to 100%.
```

## Generating a Normal Distribution

```python
# Generate random numbers using a normal distribution
bell_curve = np.random.normal(
    loc=10,
    scale=5,
    size=(4, 5)
)

print(bell_curve)
```
