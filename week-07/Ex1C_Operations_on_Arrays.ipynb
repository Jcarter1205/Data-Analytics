## Sample Arrays

```python
# Create five random sample arrays

r_1 = np.random.randint(1, 21, size=2)

r_2 = np.random.randint(1, 21, size=(1, 2))

r_3 = np.random.random(size=(2, 1))

r_4 = np.random.randint(20, 101, size=(4, 5))

r_5 = np.random.random(size=(4, 5))

# Print all five arrays
print("r_1:")
print(r_1)

print("\nr_2:")
print(r_2)

print("\nr_3:")
print(r_3)

print("\nr_4:")
print(r_4)

print("\nr_5:")
print(r_5)
```

## Permutations

```python
# Display the original r_4 array
print(r_4)
```

```python
# Shuffle the values in r_4
np.random.shuffle(r_4)

# Display the shuffled array
print(r_4)
```

The shuffle function changes the order of the rows in the array.
The values within each row stay together, but the rows move
into a different order.

```python
# Create a new array with a random permutation of r_4
new_shuffle = np.random.permutation(r_4)

print("New Shuffle:")
print(new_shuffle)

print("\nOriginal r_4:")
print(r_4)
```

Shuffle changes the original array directly, while permutation
creates a new shuffled array and leaves the original array unchanged.
I would use shuffle when I want to change the original data and
permutation when I want to keep the original data.

## Mathematical Operations

```python
# Display starting arrays and their shapes
print("r_1:")
print(r_1)
print("Shape of r_1:", r_1.shape)

print("\nr_2:")
print(r_2)
print("Shape of r_2:", r_2.shape)

# Add the two arrays
result = np.add(r_1, r_2)

print("\nAddition Result:")
print(result)
print("Result Shape:", result.shape)
```

```python
# Display starting arrays and their shapes
print("r_1:")
print(r_1)
print("Shape of r_1:", r_1.shape)

print("\nr_3:")
print(r_3)
print("Shape of r_3:", r_3.shape)

# Add the two arrays using broadcasting
result = np.add(r_1, r_3)

print("\nAddition Result:")
print(result)
print("Result Shape:", result.shape)
```

```python
# Display starting arrays and their shapes
print("r_2:")
print(r_2)
print("Shape of r_2:", r_2.shape)

print("\nr_4:")
print(r_4)
print("Shape of r_4:", r_4.shape)

# Attempt to add the arrays
result = np.add(r_2, r_4)

print(result)

# This produces a ValueError because the dimensions
# of the arrays are not compatible for broadcasting.
```

```python
# Add 1 to every value in r_4
result = np.add(r_4, np.array([1]))

print(result)
```

```python
# Add a 2D array containing 1 to r_4
result = np.add(r_4, np.array([[1]]))

print(result)
```

```python
# Add 1 to each column in r_4
result = np.add(r_4, np.array([1, 1, 1, 1, 1]))

print(result)
```

```python
# Attempt to add an array with two rows and one column
result = np.add(r_4, np.array([[1], [1]]))

print(result)

# This produces a ValueError because r_4 has 4 rows,
# but the other array has only 2 rows.
```

Broadcasting allows NumPy to combine arrays with different shapes
when their dimensions are compatible. NumPy compares the shapes
starting from the right side. If the dimensions match or one of
them is 1, NumPy can repeat values to make the operation work.
The resulting array uses the larger compatible shape.

## Rounding

```python
# Display the original random decimal values
print(r_3)
```

```python
# Remove the decimal portion without rounding
print(np.trunc(r_3))
```

```python
# Round values toward zero
print(np.fix(r_3))
```

```python
# Round values down to the nearest integer
print(np.floor(r_3))
```

```python
# Round values up to the nearest integer
print(np.ceil(r_3))
```

## Cumulative Operations

```python
# Display the array
print(r_4)
```

```python
# Calculate the sum of every value in r_4
print("Total Sum:")
print(np.sum(r_4))
```

```python
# Calculate the sum of each column
print("Sum Along Axis 0:")
print(np.sum(r_4, axis=0))
```

```python
# Calculate the sum of each row
print("Sum Along Axis 1:")
print(np.sum(r_4, axis=1))
```

```python
# Calculate the rolling total of all values
print("Cumulative Sum:")
print(np.cumsum(r_4))
```

```python
# Calculate the rolling total down each column
print("Cumulative Sum Along Axis 0:")
print(np.cumsum(r_4, axis=0))
```

## Grocery Problem

```python
# Create the first grocery bag
bag1 = np.array([2, 3, 4, 0])

# Create the second grocery bag
bag2 = np.array([4, 0, 1, 2])

print("Bag 1:")
print(bag1)

print("\nBag 2:")
print(bag2)
```

```python
# Calculate the total number of items in each bag
total_bag1 = np.sum(bag1)
total_bag2 = np.sum(bag2)

print("The first bag contains", total_bag1,
      "items, and the second bag contains", total_bag2, "items.")
```

```python
# Add the two grocery bags together
total_categories = np.add(bag1, bag2)

print("There are", total_categories[0], "vegetables,",
      total_categories[1], "fruits,",
      total_categories[2], "boxes of snacks, and",
      total_categories[3], "beverages in the two grocery bags.")
```

## Teamwork

```python
# Ali's door-knocks from Monday through Friday
ali = np.array([5, 4, 6, 4, 2])

print("Ali:")
print(ali)
```

```python
# Ben missed Monday because he was out of town
ben = np.array([0, 5, 5, 3, 4])

print("Ben:")
print(ben)
```

```python
# Carla missed Wednesday because she was sick
carla = np.array([3, 6, 0, 3, 4])

print("Carla:")
print(carla)
```

```python
# Combine Ali, Ben, and Carla into one array
all_knocks = np.array([ali, ben, carla])

print("All Door-Knocks:")
print(all_knocks)
```

```python
# Calculate each person's weekly total
person_totals = np.sum(all_knocks, axis=1)

# Calculate the group's total
group_total = np.sum(all_knocks)

print("Ali's Total:", person_totals[0])
print("Ben's Total:", person_totals[1])
print("Carla's Total:", person_totals[2])

print("Group Total:", group_total)
```

```python
# Calculate the group's total door-knocks for each day
daily_knocks = np.sum(all_knocks, axis=0)

# Calculate the rolling total of door-knocks
rolling_knocks = np.cumsum(daily_knocks)

# Estimate votes using one vote for every two door-knocks
estimated_votes = rolling_knocks * 0.5

print("Daily Door-Knocks:")
print(daily_knocks)

print("\nRolling Total of Door-Knocks:")
print(rolling_knocks)

print("\nEstimated Rolling Total of Votes:")
print(estimated_votes)
```
