# Exercise 1.A NumPy: Manipulating Arrays

```python
# Import NumPy
import numpy as np
```

## Looping Through Arrays

```python
# Create a two-dimensional array
arrA = np.array([
    [11, 12, 13],
    [14, 15, 16]
])

# Print the array
print(arrA)
```

```python
# Loop through each row and multiply by 2
for x in arrA:
    print(x * 2)
```

```python
# Loop through each individual number and multiply by 2
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop prints each row of the array multiplied by 2.
The second loop goes through the array one number at a time
and prints each number multiplied by 2.

```python
# Loop through the second dimension
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)

# The result is different because x represents the entire row,
# not the individual number y. The same row is printed repeatedly.
```

## Joining and Stacking Arrays

```python
# Join arrA with arrA along axis 0
joined_array_1 = np.concatenate((arrA, arrA), axis=0)

print(joined_array_1)
```

```python
# Join arrA with arrA along axis 1
joined_array_2 = np.concatenate((arrA, arrA), axis=1)

print(joined_array_2)
```

```python
# Stack joined_array_1 with itself
stacked_array_1 = np.stack((joined_array_1, joined_array_1))

print(stacked_array_1)
```

```python
# Try stacking arrays with different shapes
stacked_array_2 = np.stack((joined_array_1, joined_array_2))

print(stacked_array_2)

# This produces a ValueError because the two arrays
# have different shapes and cannot be stacked together.
```

```python
# Try joining along axis 1
another_2 = np.concatenate((joined_array_1, arrA), axis=1)

print(another_2)

# This produces a ValueError because the arrays have
# different numbers of rows and cannot be joined along axis 1.
```

```python
# hstack joins arrays horizontally (side by side)
horizontal_stack = np.hstack((arrA, arrA))
print("Horizontal Stack:")
print(horizontal_stack)

# vstack joins arrays vertically (one above the other)
vertical_stack = np.vstack((arrA, arrA))
print("\nVertical Stack:")
print(vertical_stack)

# dstack joins arrays along a third dimension
depth_stack = np.dstack((arrA, arrA))
print("\nDepth Stack:")
print(depth_stack)
```

## Splitting Arrays

```python
# Split joined_array_1 into 2 arrays
splitsky = np.split(joined_array_1, 2)

print(splitsky)

# This creates 2 arrays, each with 2 rows and 3 columns.
```

```python
# Split joined_array_1 into 3 arrays
spliterino = np.array_split(joined_array_1, 3)

print(spliterino)

# This creates 3 arrays. Each is two-dimensional.
# The first has 2 rows, and the other two have 1 row each.
```

## Searching and Replacing Array Values

```python
# Find even numbers in arrA and replace odd numbers with 'odd'
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')

print(result_1)
```

## Sorting and Filtering Arrays

```python
# Create a two-dimensional array
arrB = np.array([
    [7, 6],
    [13, 5],
    [7, 3]
])

# Sort the values in each row
arrB_sorted = np.sort(arrB)

print(arrB_sorted)
```

```python
# Display the original array
print("Original Array:")
print(arrB)

# Find values less than 7
z = arrB < 7

# Store the filtered values
arrC = arrB[z]

print("\nValues Less Than 7:")
print(arrC)
```

```python
# Find values greater than or equal to 7
y = arrB >= 7

# Store the filtered values
arrD = arrB[y]

print("Values Greater Than or Equal to 7:")
print(arrD)
```
