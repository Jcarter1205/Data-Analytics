# Exercise 2.C Creating and Calling Functions
## Lab 3: Payroll Tax Functions

This lab demonstrates how to create custom Python functions to calculate Social Security, Medicare, and federal income tax withholding. It uses parameters to accept gross pay and withholding codes, return calculated tax amounts, and apply different tax rates using conditional statements.

```python
def get_soc_sec_tax(gross_pay):
    tax = gross_pay * 0.062
    return tax
```

```python
def get_medicare_tax(gross_pay):
    tax = gross_pay * 0.0145
    return tax
```

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        tax_rate = 0.23
    elif withholding_code == 1:
        tax_rate = 0.21
    elif withholding_code == 2:
        tax_rate = 0.195
    elif withholding_code == 3:
        tax_rate = 0.185
    else:
        tax_rate = 0.18 - ((withholding_code - 4) * 0.005)

    tax = gross_pay * tax_rate
    return tax
```

```python
gross_pay = 750
withholding_code = 0

print("Person 1")
print(f"Gross Pay: ${gross_pay:.2f}")
print(f"Social Security Tax: ${get_soc_sec_tax(gross_pay):.2f}")
print(f"Medicare Tax: ${get_medicare_tax(gross_pay):.2f}")
print(f"Federal Tax: ${get_federal_tax(gross_pay, withholding_code):.2f}")
```
