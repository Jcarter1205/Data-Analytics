# Exercise 2.C Creating and Calling Functions
## Lab 1: Working with Custom Functions

This lab demonstrates how to define custom Python functions, use parameters to provide information, and return formatted text.

```python
def favorite_things(name, movie, music):
    return f"My name is {name}.\nMy favorite movie is {movie}.\nMy favorite musician or band is {music}."
```

```python
print(favorite_things("Jada Carter", "The Devil Wears Prada", "SZA"))
```

```python
def why_im_here(purpose):
    return purpose
```

```python
print(why_im_here("I joined the Year Up United Data Analyst program to develop my technical skills, gain hands-on experience, and prepare for a career in data analytics."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("New York City", "there is always something new to explore"))
```

```python
print(favorite_place("Paris", "I love the architecture, fashion, and atmosphere"))
```
