# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(part_code):
    colon_position = part_code.find(':')
    return part_code[:colon_position]
```

```python
get_supplier_code('ACME:123-L')
```

```python
def get_product_num(part_code):
    colon_position = part_code.find(':')
    hyphen_position = part_code.find('-')

    return part_code[colon_position + 1:hyphen_position]
```

```python
get_product_num('ACME:123-L')
```

```python
def get_size(part_code):
    hyphen_position = part_code.find('-')

    return part_code[hyphen_position + 1:]
```

```python
get_size('ACME:123-L')
```

```python
part_code1 = 'ACME:123-L'
part_code2 = 'DI:12345-M'
part_code3 = 'ACE:1-12'
```

```python
print(f"For {part_code1}, the supplier is {get_supplier_code(part_code1)}. Part number {get_product_num(part_code1)} is available in size {get_size(part_code1)}.")
```

```python
print(f"For {part_code2}, the supplier is {get_supplier_code(part_code2)}. Part number {get_product_num(part_code2)} is available in size {get_size(part_code2)}.")
```

```python
print(f"For {part_code3}, the supplier is {get_supplier_code(part_code3)}. Part number {get_product_num(part_code3)} is available in size {get_size(part_code3)}.")
```

```python
part_codes = [part_code1, part_code2, part_code3]
```

```python
for code in part_codes:
    print(f"For {code}, the supplier is {get_supplier_code(code)}. Part number {get_product_num(code)} is available in size {get_size(code)}.")
```
